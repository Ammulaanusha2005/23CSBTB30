```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import zipfile
import io
import time
import os

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
from sklearn.decomposition import PCA

import warnings
warnings.filterwarnings("ignore")

sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
url = "https://archive.ics.uci.edu/static/public/352/online+retail.zip"

print("Downloading Online Retail dataset...")

response = requests.get(url, timeout=180)
response.raise_for_status()

print("Download completed!")

zip_file = zipfile.ZipFile(io.BytesIO(response.content))

excel_files = [
    name for name in zip_file.namelist()
    if name.lower().endswith(".xlsx")
]

if not excel_files:
    raise FileNotFoundError(
        "Excel dataset not found inside downloaded ZIP"
    )

print("Dataset file:", excel_files[0])

with zip_file.open(excel_files[0]) as f:
    df = pd.read_excel(f, engine="openpyxl")

df.columns = df.columns.str.strip()

print("\nDATASET LOADED SUCCESSFULLY")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
print("DATASET INFORMATION")

print("\nShape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isna().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nUnique Customers:")
print(df["CustomerID"].nunique())

print("\nUnique Countries:")
print(df["Country"].nunique())

print("\nStatistical Summary:")
display(df.describe())

print("\nInspection completed!")
```

```python
country_counts = df["Country"].value_counts().head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=country_counts.values,
    y=country_counts.index
)

plt.title("Top 10 Countries in Online Retail Dataset")
plt.xlabel("Number of Transactions")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

df["TotalAmount"] = (
    df["Quantity"] * df["UnitPrice"]
)

monthly_sales = (
    df.groupby(df["InvoiceDate"].dt.month)["TotalAmount"]
    .sum()
)

plt.figure(figsize=(10, 5))

monthly_sales.plot(marker="o")

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Sales Amount (£)")
plt.xticks(range(1, 13))

plt.tight_layout()
plt.show()
```

```python

clean_df = df.copy()

clean_df = clean_df.dropna(subset=["CustomerID"])

clean_df["CustomerID"] = clean_df["CustomerID"].astype(int)

clean_df = clean_df[
    ~clean_df["InvoiceNo"].astype(str).str.startswith("C")
]

clean_df = clean_df[clean_df["Quantity"] > 0]

clean_df = clean_df[clean_df["UnitPrice"] > 0]

clean_df = clean_df.dropna(subset=["InvoiceDate"])

clean_df = clean_df.drop_duplicates()

clean_df["TotalAmount"] = (
    clean_df["Quantity"] * clean_df["UnitPrice"]
)

print("DATA CLEANING COMPLETED")

print("Original records:", len(df))
print("Cleaned records:", len(clean_df))
print("Remaining customers:", clean_df["CustomerID"].nunique())

display(clean_df.head())
```

```python

reference_date = (
    clean_df["InvoiceDate"].max() +
    pd.Timedelta(days=1)
)

rfm = clean_df.groupby("CustomerID").agg(
    Recency=(
        "InvoiceDate",
        lambda x: (reference_date - x.max()).days
    ),

    Frequency=(
        "InvoiceNo",
        "nunique"
    ),

    Monetary=(
        "TotalAmount",
        "sum"
    )
)

rfm = rfm.replace([np.inf, -np.inf], np.nan)
rfm = rfm.dropna()

rfm = rfm[
    (rfm["Recency"] >= 0) &
    (rfm["Frequency"] > 0) &
    (rfm["Monetary"] > 0)
]

print("RFM CALCULATION COMPLETED")

print("Total customers:", len(rfm))

display(rfm.head(10))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(
    rfm["Recency"],
    bins=30,
    ax=axes[0]
)
axes[0].set_title("Customer Recency")

sns.histplot(
    rfm["Frequency"],
    bins=30,
    ax=axes[1]
)
axes[1].set_title("Customer Frequency")

sns.histplot(
    rfm["Monetary"],
    bins=30,
    ax=axes[2]
)
axes[2].set_title("Customer Monetary")

plt.tight_layout()
plt.show()
```

```python

rfm_log = np.log1p(
    rfm[["Recency", "Frequency", "Monetary"]]
)

scaler = StandardScaler()

scaled_data = scaler.fit_transform(rfm_log)

print("DATA STANDARDIZATION COMPLETED")

print("Scaled shape:", scaled_data.shape)

print("First five records:")
print(scaled_data[:5])
```

```python
wcss = []

for k in range(2, 10):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(scaled_data)

    wcss.append(model.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 10),
    wcss,
    marker="o"
)

plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")

plt.xticks(range(2, 10))
plt.show()
```

```python
silhouette_results = []

for k in range(2, 10):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(scaled_data)

    score = silhouette_score(
        scaled_data,
        labels,
        sample_size=min(2000, len(scaled_data)),
        random_state=42
    )

    silhouette_results.append(score)

    print(
        "K =", k,
        "Silhouette Score =", round(score, 4)
    )

plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 10),
    silhouette_results,
    marker="o"
)

plt.title("Silhouette Analysis")
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")

plt.show()
```

```python
k = 4

kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

start = time.time()

kmeans_labels = kmeans.fit_predict(scaled_data)

kmeans_time = time.time() - start

rfm["KMeans_Cluster"] = kmeans_labels

print("K-MEANS COMPLETED")

print("Execution time:", round(kmeans_time, 4), "seconds")

print("\nCustomer distribution:")

print(
    rfm["KMeans_Cluster"]
    .value_counts()
    .sort_index()
)
```

```python

sample_size = min(2000, len(rfm))

sample_indices = np.random.RandomState(42).choice(
    len(rfm),
    size=sample_size,
    replace=False
)

X_sample = scaled_data[sample_indices]


hierarchical = AgglomerativeClustering(
    n_clusters=k,
    linkage="ward"
)

start = time.time()

hierarchical_labels = hierarchical.fit_predict(X_sample)

hierarchical_time = time.time() - start

print("HIERARCHICAL CLUSTERING COMPLETED")

print("Execution time:", round(hierarchical_time, 4), "seconds")

print("\nCluster distribution:")

print(pd.Series(hierarchical_labels).value_counts().sort_index())
```

```python
# K-Means predictions for same sample
kmeans_sample_labels = kmeans.predict(X_sample)


def calculate_metrics(name, X, labels, execution_time):

    return {
        "Algorithm": name,

        "Silhouette Score": silhouette_score(
            X, labels
        ),

        "Davies-Bouldin Index": davies_bouldin_score(
            X, labels
        ),

        "Calinski-Harabasz Score": calinski_harabasz_score(
            X, labels
        ),

        "Execution Time": execution_time
    }

results = []

results.append(
    calculate_metrics(
        "K-Means",
        X_sample,
        kmeans_sample_labels,
        kmeans_time
    )
)

results.append(
    calculate_metrics(
        "Hierarchical",
        X_sample,
        hierarchical_labels,
        hierarchical_time
    )
)

comparison_df = pd.DataFrame(results)

print("ALGORITHM COMPARISON")

display(comparison_df.round(4))
```

```python
pca = PCA(n_components=2)

pca_result = pca.fit_transform(scaled_data)

rfm["PCA1"] = pca_result[:, 0]
rfm["PCA2"] = pca_result[:, 1]

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=rfm,
    x="PCA1",
    y="PCA2",
    hue="KMeans_Cluster",
    palette="viridis",
    s=30
)

plt.title("Customer Segmentation Using K-Means")

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.legend(title="Cluster")

plt.show()
```

```python
cluster_profile = rfm.groupby(
    "KMeans_Cluster"
).agg({

    "Recency": "mean",

    "Frequency": "mean",

    "Monetary": "mean"

})

# Count the number of customers in each cluster using the index
cluster_profile["Customer_Count"] = rfm.groupby("KMeans_Cluster").size()

print("CUSTOMER CLUSTER PROFILE")

display(cluster_profile.round(2))
```

```python
median_r = rfm["Recency"].median()
median_f = rfm["Frequency"].median()
median_m = rfm["Monetary"].median()

def customer_type(row):

    if (
        row["Recency"] <= median_r
        and row["Frequency"] >= median_f
        and row["Monetary"] >= median_m
    ):
        return "Champions"

    elif (
        row["Recency"] <= median_r
        and row["Frequency"] >= median_f
    ):
        return "Loyal Customers"

    elif row["Monetary"] >= median_m:
        return "Big Spenders"

    elif (
        row["Recency"] > median_r
        and row["Frequency"] < median_f
    ):
        return "At Risk Customers"

    else:
        return "Regular Customers"


# Assign customer names based on cluster averages
cluster_names = {}

for cluster, row in cluster_profile.iterrows():

    cluster_names[cluster] = customer_type(row)

rfm["Customer_Segment"] = rfm["KMeans_Cluster"].map(
    cluster_names
)

print("CUSTOMER SEGMENTS")

display(
    rfm[
        ["Recency", "Frequency", "Monetary",
         "KMeans_Cluster", "Customer_Segment"]
    ].head(15)
)

print("\nSegment distribution:")
print(rfm["Customer_Segment"].value_counts())
```

```python
plt.figure(figsize=(9, 5))

segment_counts = rfm["Customer_Segment"].value_counts()

plt.pie(
    segment_counts.values,
    labels=segment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Segment Distribution")

plt.show()
```

```python
sensitivity_results = []

for k in range(2, 9):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(scaled_data)

    sample_labels = labels[sample_indices]

    sensitivity_results.append({

        "K": k,

        "Silhouette Score": silhouette_score(
            X_sample,
            sample_labels
        ),

        "Davies-Bouldin Index": davies_bouldin_score(
            X_sample,
            sample_labels
        ),

        "Calinski-Harabasz Score":
            calinski_harabasz_score(
                X_sample,
                sample_labels
            )
    })

sensitivity_df = pd.DataFrame(sensitivity_results)

display(sensitivity_df.round(4))
```

```python
feature_sets = {

    "Recency": ["Recency"],

    "Frequency": ["Frequency"],

    "Monetary": ["Monetary"],

    "Recency + Frequency":
        ["Recency", "Frequency"],

    "Recency + Monetary":
        ["Recency", "Monetary"],

    "Frequency + Monetary":
        ["Frequency", "Monetary"],

    "All RFM":
        ["Recency", "Frequency", "Monetary"]
}

ablation_results = []

for name, columns in feature_sets.items():

    X = np.log1p(rfm[columns])

    X = StandardScaler().fit_transform(X)

    model = KMeans(
        n_clusters=4,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    ablation_results.append({

        "Features": name,

        "Silhouette Score": silhouette_score(
            X,
            labels,
            sample_size=min(2000, len(X)),
            random_state=42
        ),

        "Davies-Bouldin Index": davies_bouldin_score(
            X,
            labels
        )
    })

ablation_df = pd.DataFrame(ablation_results)

display(ablation_df.round(4))
```
